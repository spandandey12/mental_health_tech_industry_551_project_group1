## Contributions and Reflection

### How the team worked together

Our team was able to complete the project and deliver the final dashboard on time. Communication among active members was generally effective when discussing ideas, dividing tasks, and reviewing progress. Most of the project development and final preparation were completed through coordination between the more involved team members.

### My role in the group

I contributed to the project by helping design and implement parts of the dashboard, supporting the data analysis and visualization process, and coordinating task progress. I also helped review the dashboard functionality, checked whether the project requirements were met, and made improvements before the final presentation.

### Roles of other team members

**Manpreet Singh** had limited involvement in the project overall. His contribution to preparation, communication, and final presentation support was lower compared with the other team members. During the presentation stage, his participation was minimal, and he was not well prepared to answer questions about the presentation.

**Spandan Dey** contributed to data preparation and integration within the dashboard. He also assisted with testing the application and helped improve the usability and presentation of the final product.

### What went well

- The core project tasks were completed successfully.
- Communication between active contributors was effective.
- The dashboard was finished and prepared for presentation.
- The team was able to manage the workload and submit the project on time.

### What could be improved

A major area for improvement was more balanced participation across all team members. In particular, stronger involvement from every member during both the development process and presentation preparation would have improved the overall teamwork experience. Better accountability and clearer expectations for each person’s responsibilities would also help future group work run more smoothly.

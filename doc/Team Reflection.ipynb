## Contributions and Reflection

### How the team worked together

Our team collaborated effectively throughout the project. We maintained regular communication when discussing ideas, dividing responsibilities, and reviewing progress. Each member contributed to the development of the dashboard and supported the group when challenges appeared. Overall, the teamwork was smooth and productive.

### My role in the group

I contributed to the project by helping design and implement parts of the dashboard and supporting the data analysis and visualization process. I also helped coordinate tasks and ensured that the project requirements were clearly understood. In addition, I participated in reviewing the dashboard functionality and making improvements based on feedback.

### Roles of other team members

**Manpreet Singh** contributed to the dashboard implementation and worked on visualization components and layout improvements. He actively participated in discussions and helped ensure that the dashboard features worked correctly.

**Spandan Dey** supported data preparation and integration into the dashboard. He also helped test the application and provided suggestions for improving usability and presentation.

### What went well

- Team members communicated effectively throughout the project.
- Responsibilities were shared fairly across the team.
- The team actively incorporated feedback from teaching assistants and peers.
- The overall collaboration environment was positive and supportive.

### What could be improved

One challenge was coordinating meeting times due to different schedules. However, we managed this by communicating online and keeping everyone updated. Overall, the project process went smoothly and the collaboration worked well.
